```python
def game_number(number: int = 1) -> int:
    """Угадываем число с помощью деления нового промежутка пополам"""
    #Начальное значение счетчика попыток
    count = 0 
    #Задаем рандомно начального значения отсчета (если нужно сократить колическтво попыток на поиск, лучше поставить значение predict = 50)
    predict = np.random.randint(1, 101)
    #prtdict = 50 #Если начальную позицию загадывать не рандомно 
    #Верхний диапазон заданой области
    max_bound = 100 
    #Нижний диапазон заданой области
    min_bound = 1
    # Проверяем угадали мы число или нет
    while number != predict:
        # Увеличиваем счетчик при каждой попытке угадывания
        count += 1
        # Если число больше предложенного нами
        if number > predict:
            # Присваиваем нижней границе диапазона новое значение, +1 потому что число точно больше predict
            min_bound = predict + 1
            # Делим диапазон нашего поиска на 2 
            predict = (max_bound + min_bound) // 2
        # Если число меньше предложенного нами     
        elif number < predict:
            # Присваиваем верхней границе диапазона новое значение, -1 потому что число точно меньше predict
            max_bound = predict - 1
            # Делим диапазон нашего поиска на 2 
            predict = (max_bound + min_bound) // 2

    return count  
```
